## 🎬 Challenge Solution: Will We Go to the Movies? (Our Own Weights and Bias)

We use a **single-layer perceptron** with three yes/no inputs:

- **x₁ (weather):** 1 = good weather, 0 = bad weather
- **x₂ (company):** 1 = a friend is coming, 0 = going alone
- **x₃ (proximity):** 1 = the theater is nearby, 0 = it is far

The output is **1 = go to the movies** or **0 = stay home**.

### ⚙️ Our Own Weights and Bias

| Input | Weight | Meaning |
|---|---|---|
| x₁ weather | **2.0** | Most important: we really don't want to go in bad weather |
| x₂ company | **1.5** | Important: movies are more fun with friends |
| x₃ proximity | **1.0** | Helpful, but a nearby theater is not essential |
| bias | **-3.0** | The "bar" the total must pass before we decide to go |

We use the **same step activation function** as in the examples above:

$$
z = 2.0\,x_1 + 1.5\,x_2 + 1.0\,x_3 - 3.0
\qquad
\text{output} =
\begin{cases}
1 & \text{if } z > 0 \\
0 & \text{otherwise}
\end{cases}
$$

```python
from itertools import product

# Inputs: weather (x1), company (x2), proximity (x3)
weights = [2.0, 1.5, 1.0]   # w1, w2, w3  (our own values)
bias = -3.0                 # b

def step(z):
    return 1 if z > 0 else 0

print("weather company near |   z   | Go to movies?")
print("---------------------------------------------")
for x in product([0, 1], repeat=3):
    z = sum(w * xi for w, xi in zip(weights, x)) + bias
    output = step(z)
    answer = "YES" if output == 1 else "no"
    print(f"   {x[0]}       {x[1]}      {x[2]}   | {z:>5.1f} |   {output} ({answer})")
```

### 🧠 Interpretation

- **Output 1** → we go to the movies. **Output 0** → we stay home.
- We go when the weather is good **and** a friend comes (even if the theater is far).
- Good weather plus a nearby theater but no company gives z = 0. Since the step function needs z **greater than** 0, we stay home.
- Good company plus a nearby theater but bad weather is not enough. We stay home.
- Everything good (1, 1, 1) gives the highest score, z = 1.5, so we go.

### 💬 Talking Points

1. **Weights show what matters most.** Weather has the biggest weight (2.0), so it has the biggest influence on the decision. Changing the weights changes the decision without changing the code.
2. **The bias sets how hard it is to say "yes".** With a bias of -3.0, the weighted inputs must add up to more than 3 before the neuron fires. Making the bias closer to 0 would make us go to the movies more easily.
3. **Unlike the earlier example (all weights 1.0, bias -2.5), our perceptron needs no single input to be perfect.** Different combinations of inputs can lead to "go", because each input has a different importance. This is how a perceptron mimics real decision making.
